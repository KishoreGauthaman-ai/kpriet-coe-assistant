```python
class node:
    def __init__(self, data):
        self.data = data
        self.next = None
    def insert(self,position):
        if position == 0:
            new_node = node(self.data)
            new_node.next = self.next
            self.data = position
            self.next = new_node
        else:
            if self.next is None:
                self.next = node(position)
            else:
                self.next.insert(position)
```
