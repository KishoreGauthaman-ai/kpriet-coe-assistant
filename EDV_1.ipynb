```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('Mall_Customers.csv')
data.head()
```

```python
data.info()
```

### Adding the date column for the following line chart (date vs spending score)

```python
data["date"] = pd.date_range(start='01-01-2025', periods=len(data), freq='D')
```

```python
data.head()
```

```python
plt.plot(data['date'], data['Spending Score (1-100)'])
plt.xlabel('Date')
plt.ylabel('Spending Score (1-100)')
plt.title('Spending Score Over Time')
plt.show()
```

## <b>Interpretation</b>
        The people doesn't shown much interest in buying Things in Mall..


```python
sns.scatterplot(x=data['Age'],y=data['Spending Score (1-100)'], color='Deepskyblue')
plt.show()
```

## <b>Interpretation</b>
        The people in age between 20 to 40 are spending more and the people above 40 spending less.

```python
sns.countplot(x=data['Gender'],hue=data['Gender'])
```

## <b>Interpretation</b>
        Female are using the mall more than Male

```python
sns.barplot(x=data['Gender'],y=data['Spending Score (1-100)'],hue=data['Gender'])
```

## <b>Interpretation</b>
        Since the Female count more so the spending of the female are also more.
        But the count of male is less compare to female but the spending od the male is nearly equal to female.
