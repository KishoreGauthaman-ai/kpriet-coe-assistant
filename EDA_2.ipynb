```python
import pandas as pd
import seaborn as sns
```

```python
import os
os.getcwd()
```

```python
data = pd.read_csv('Mall_Customers.csv')
```

```python
data.head() 
```

```python
sns.boxplot(data = data[['Age','Spending Score (1-100)']], palette=['yellow','green'])
```

```python
sns.violinplot(x = data['Gender'], y = data['Spending Score (1-100)'], split=True, palette=['yellow','green'])   
```

```python
sns.histplot(data = data['Age'], bins=10, kde=True)
```

```python
sns.kdeplot(data = data['Age'], fill=True, color='yellow', alpha=0.5)
```

```python
sns.pairplot(data, hue='Gender')
```

```python
sns.barplot(x = data['Gender'], y = data['Spending Score (1-100)'], hue=data['Gender'], palette=['yellow','green'])
```

```python
sns.countplot(x = data['Gender'], palette=['yellow','green'])
```
